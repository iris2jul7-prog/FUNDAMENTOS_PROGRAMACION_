Programa de tabla de pitagoras de 10x10 utilizando listas y ciclos respetando un rango ,en base a  renglones y columnas para sacar resultados entre si, y poder tener resultados numericos reales sin necesidad de multiplicacion

PARTE A: Construir la matriz

```python
# construir la matriz
tabla = []
for fila in range(1, 11):
    nueva_fila = []
    valor = 0
    for columna in range(1, 11):
        valor += fila
        nueva_fila.append(valor)
    tabla.append(nueva_fila)
```

PARTE B:Imprimir la tabla

```python
#imprimir tabla
def imprimir_tabla(tabla_matriz):

    print("", end="")
    for columna in range(1, 11):
        print(f"{columna:3d}", end="\t")
    print()

    for fila in range(10):
        print(f"{fila + 1:3d}", end="\t")

        for columna in range(10):
            print(f"{tabla_matriz[fila][columna]:3d}", end="\t")
        print()
```

PARTE C: Consultar el producto

```python
#consutar producto
def consultar_producto(tabla, renglon, columna):
    return tabla[renglon - 1][columna - 1]
```

PARTE D: cantidades y resultados

```python
# Pedir cantidades y dar resultados
imprimir_tabla(tabla)

renglon = int(input("Ingresa el renglon/numero que te gustaria multiplicar del 1-10 "))
while renglon < 1 or renglon > 10:
    print("Numero invalido , debe ser entre 1 y 10")
    renglon = int(input("ingresa el renglon/numero que te gustaria multiplicar 1-10: "))

columna = int(input("Ingresa la columna/numero por el que se multiplicara 1-10: "))
while columna < 1 or columna > 10:
    print("Numero invalido , debe ser entre 1 y 10")
    columna = int(input("Ingresa la columna/numero por el que se multiplicara  1-10: "))
#resultado
resultado = consultar_producto(tabla, renglon, columna)
print(f"El resultado de {renglon} por {columna} es: {resultado}")
```
