```python
#Ejercicio 1
dias = ("Lunes", "Martes", "Miercoles", "Jueves", "Viernes", "Sabado", "Domingo")
km = (120, 85, 0, 210, 95, 0, 150)

def calcular_rodada(datos_km):
    total = 0
    dias_con_rodada = 0

    for km in datos_km:
        if km > 0:
            total += km
            dias_con_rodada += 1

    if dias_con_rodada > 0:
        promedio = total / dias_con_rodada
    else:
        promedio = 0.0

    return total, dias_con_rodada, promedio

total, dias_rodados, promedio = calcular_rodada(km)

max_km = max(km)
indice_max = km.index(max_km)

dia_mas_rodada = dias[indice_max]

print(f"Total: {total} km")
print(f"Días rodados: {dias_rodados}")
print(f"Promedio por día rodado: {promedio:.1f} km")
print(f"El día con más rodada fue el :{dia_mas_rodada} con {max_km} km")
```

```python
#Ejercicio 2
votos = [("Metallica", 48), ("Slayer", 22), ("Megadeth", 35), ("Anthrax", 15), ("Iron Maiden", 40)]

banda_ganadora = ""
max_votos = -1

for banda, cantidad in votos:
    if cantidad > max_votos:
        max_votos = cantidad
        banda_ganadora = banda

print(f"Ganador: {banda_ganadora} con 48 votos")

def promedio_votos(lista_votos):
    total_votos = 0
    for _, cantidad in lista_votos:
        total_votos += cantidad
    return total_votos / len(lista_votos) if lista_votos else 0

promedio = promedio_votos(votos)
print(f"Promedio: {promedio:.1f}")

print("Arriba del promedio:")
for banda, cantidad in votos:
    if cantidad > promedio:
      print(f"{banda} ({cantidad})")
```

```python
#Ejercicio 11
def rendimiento():
    while True:
        try:
            km_str = input("Kilómetros recorridos: ")
            litros_str = input("Litros cargados: ")

            km = float(km_str)
            litros = float(litros_str)

            if litros == 0:
                print("Los litros no pueden ser 0")
                continue

            rend = round(km / litros, 1)
            return rend

        except ValueError:
            print("Escribe solo números")

resultado = rendimiento()
print(f"Rendimiento: {resultado} km/l")

if resultado >= 20:
    print("Excelente")
elif resultado >= 15:
    print("Normal")
else:
    print("Revisa tu moto")
```

```python
#Ejercicio 14
sala = [["X", "L", "L"], ["L", "X", "L"], ["X", "X", "L"]]

def reservar(sala):
    try:
        fila_str = input("Fila (0-2): ")
        col_str = input("Columna (0-2): ")

        fila = int(fila_str)
        col = int(col_str)

        if fila < 0 or col < 0:
            raise IndexError

        asiento = sala[fila][col]

        if asiento == "X":
            print("Ocupado, elige otro")
            return False
        else:
            sala[fila][col] = "X"
            print("Reservado, disfruta")
            return True

    except ValueError:
        print("Escribe números")
        return False
    except IndexError:
        print("Ese asiento no existe")
        return False

no_reserva= True
while no_reserva:
    if reservar(sala):
        no_reserva = False

for fila in sala:
    for asiento in fila:
        print(asiento, end=" ")
    print()
```

```python
#Ejercicio 15
cupones = {"METAL10": 10, "MOTO20": 20, "ROCK15": 15}

def aplicar(total, codigo):
    try:
        codigo_limpio = codigo.strip().upper()
        descuento = cupones[codigo_limpio]
        total_con_descuento = total * (1 - descuento / 100)
        return total_con_descuento
    except KeyError:
        print("Cupón no válido")
        return total

total_inicial = 1200
codigos_prueba = [" moto20 ", "GRATIS", "metal10"]

for cod in codigos_prueba:
    nuevo_total = aplicar(total_inicial, cod)
    print(f"{cod} : ${nuevo_total}")
```

```python
#Ejercicio 16
def evaluar(contrasena):
    fallas = []


    if len(contrasena) < 8:
        fallas.append('mínimo 8 caracteres')

    tiene_numero = False
    for char in contrasena:
        if char.isdigit():
            tiene_numero = True
            break
    if not tiene_numero:
        fallas.append('falta un número')


    tiene_mayuscula = False
    for char in contrasena:
        if char.isupper():
            tiene_mayuscula = True
            break
    if not tiene_mayuscula:
        fallas.append('falta una mayúscula')


    tiene_espacio = False
    for char in contrasena:
        if char == ' ':
            tiene_espacio = True
            break
    if tiene_espacio:
        fallas.append('sin espacios')

    return fallas

pruebas = ["metal", "Harley2008", "road glide 08", "KLR650moto"]
for p in pruebas:
    resultado = evaluar(p)
    if len(resultado) == 0:
        print(f"{p}: segura")
    else:
        print(f"{p}: {resultado}")
```

```python
#Ejercicio 19
def usuario(nombre, usados):
    partes = nombre.lower().split()
    inicial_nombre = partes[0][0]
    apellido = partes[1]
    base_user = inicial_nombre + apellido

    user_final = base_user
    contador = 2

    while user_final in usados:
        user_final = base_user + str(contador)
        contador += 1

    usados.append(user_final)
    return user_final

alumnos = ["Bruce Dickinson", "Brian Dickinson", "Ana Lopez", "Andrea Lopez", "Angel Lopez"]
usuarios_nuevos = []

for alumno in alumnos:
    user = usuario(alumno, usuarios_nuevos)
    print(f"{alumno} → {user}")
```

```python
#Ejercicio 20
prohibidas = ["tonto", "menso", "baboso"]

def censurar(frase):
    palabras = frase.split()
    contador = 0

    for i in range(len(palabras)):

        palabra_limpia = palabras[i].lower()
        if palabra_limpia in prohibidas:
            palabras[i] = "*" * len(palabras[i])
            contador += 1

    frase_censurada = " ".join(palabras)
    return (frase_censurada, contador)

pruebas = [
    "no seas menso y pásame el código",
    "Tonto el que no haga commit",
    "todo bien profe"]

for p in pruebas:
    frase_final, total_censuradas = censurar(p)
    print(f"{frase_final} ({total_censuradas} censuradas)")
```

```python
#Ejercicio 21
def cobro(entrada, salida):
    horas = salida - entrada
    if horas <= 1:
        return 20
    else:
        return 20 + (horas - 1) * 15

autos = [("ABC123", 9, 11), ("XYZ789", 8, 15), ("MOT650", 10, 11), ("QRO001", 13, 13)]

cobros_dict = {}

for placa, entrada, salida in autos:
    horas = salida - entrada
    monto = cobro(entrada, salida)
    cobros_dict[placa] = monto
    print(f"{placa}: {horas} h → ${monto}")

total_dia = sum(cobros_dict.values())
print(f"Total del día: ${total_dia}")
```

```python
#Ejercicio 22


partidos = (("Pumas", "Tigres", 2, 1), ("América", "Pumas", 0, 0), ("Tigres", "América", 3, 2),
            ("Pumas", "América", 1, 2), ("Tigres", "Pumas", 1, 1))

puntos = {}

for local, visitante, goles_local, goles_vis in partidos:
    if local not in puntos:
        puntos[local] = 0
    if visitante not in puntos:
        puntos[visitante] = 0

    if goles_local > goles_vis:
        puntos[local] += 3
    elif goles_local < goles_vis:
        puntos[visitante] += 3
    else:
        puntos[local] += 1
        puntos[visitante] += 1

print(puntos)


campeon = None
max_puntos = -1
for equipo, pts in puntos.items():
    if pts > max_puntos:
        max_puntos = pts
        campeon = equipo

print(f"Campeón: {campeon} con {puntos[campeon]} puntos")
```

```python
#Ejercicio 23
saldos = {"Ana": 500, "Luis": 200, "Mía": 300}
movimientos = ["Ana:-150", "Luis:-250", "Mía:+100", "Pedro:-50", "Ana:-cien", "Mía:-380", "Luis:-200"]

errores = 0

for mov in movimientos:
    try:
        nombre, monto_str = mov.split(":")
        monto = int(monto_str)

        if saldos[nombre] + monto < 0:
            print(f"{nombre}: saldo insuficiente")
        else:
            saldos[nombre] += monto

    except ValueError:
        print(f"Monto inválido en '{mov}'")
        errores += 1
    except KeyError:
        print(f"{nombre} no tiene pulsera")
        errores += 1

print(saldos)
print(f"movimientos con error: {errores}")
```

```python
#Ejercicio 24
nombres = ("Ana", "Luis", "Mía", "Beto")
calif = [[85, 90, 78, 92], [60, 75, 70, 68], [95, 98, 100, 97], [70, 65, 80, 75]]

def promedio(lista):
    return sum(lista) / len(lista)

promedios_alumnos = {}
for i in range(len(nombres)):
    prom_alumno = promedio(calif[i])
    promedios_alumnos[nombres[i]] = prom_alumno
    print(f"{nombres[i]}: {prom_alumno}")


num_parciales = len(calif[0])
for j in range(num_parciales):
    columna_parcial = []
    for fila in calif:
        columna_parcial.append(fila[j])
    print(f"Parcial {j + 1}: {promedio(columna_parcial)}")


mejor_alumno = nombres[0]
for alumno in nombres:
    if promedios_alumnos[alumno] > promedios_alumnos[mejor_alumno]:
        mejor_alumno = alumno

print(f"Mejor alumno: {mejor_alumno}")
```

```python
#Ejercicio 25
playlist = ["Master of Puppets - Metallica - 8:35","Holy Wars - Megadeth - 6:32","Fade to Black - Metallica - 6:57",
"The Trooper - Iron Maiden - 4:10",
    "Hangar 18 - Megadeth - 5:14",
    "Aces High - Iron Maiden - 4:31"]

def a_segundos(tiempo_str):
    minutos, segundos = tiempo_str.split(":")
    return int(minutos) * 60 + int(segundos)

def formato(segundos):
    minutos = segundos // 60
    seg = segundos % 60
    return f"{minutos}:{seg:02d}"

tiempo_artistas = {}
total_segundos = 0

for cancion in playlist:
    nombre, artista, duracion_str = cancion.split(" - ")
    segundos = a_segundos(duracion_str)

    if artista not in tiempo_artistas:
        tiempo_artistas[artista] = 0
    tiempo_artistas[artista] += segundos

    total_segundos += segundos

for artista, segs in tiempo_artistas.items():
    print(f"{artista}: {formato(segs)}")

mas_tiempo = None
max_segundos = -1

for artista, segs in tiempo_artistas.items():
    if segs > max_segundos:
        max_segundos = segs
        mas_tiempo = artista

print(f"Más tiempo: {mas_tiempo}")
print(f"Total: {formato(total_segundos)}")
```
