```python
#Ejercicio 1
dias = ("Lunes", "Martes", "Miercoles", "Jueves", "Viernes", "Sabado", "Domingo")
km = (120, 85, 0, 210, 95, 0, 150)

def calcular_rodada(datos_km):
    total = 0
    dias_con_rodada = 0

    for km in datos_km:
        if km > 0:
            total += km
            dias_con_rodada += 1

    if dias_con_rodada > 0:
        promedio = total / dias_con_rodada
    else:
        promedio = 0.0

    return total, dias_con_rodada, promedio

total, dias_rodados, promedio = calcular_rodada(km)

max_km = max(km)
indice_max = km.index(max_km)

dia_mas_rodada = dias[indice_max]

print(f"Total: {total} km")
print(f"Días rodados: {dias_rodados}")
print(f"Promedio por día rodado: {promedio:.1f} km")
print(f"El día con más rodada fue el :{dia_mas_rodada} con {max_km} km")
```

```python
#Ejercicio 2
votos = [("Metallica", 48), ("Slayer", 22), ("Megadeth", 35), ("Anthrax", 15), ("Iron Maiden", 40)]

banda_ganadora = ""
max_votos = -1

for banda, cantidad in votos:
    if cantidad > max_votos:
        max_votos = cantidad
        banda_ganadora = banda

print(f"Ganador: {banda_ganadora} ")

def promedio_votos(lista_votos):
    total_votos = 0
    for _, cantidad in lista_votos:
        total_votos += cantidad
    return total_votos / len(lista_votos) if lista_votos else 0

promedio = promedio_votos(votos)
print(f"Promedio: {promedio:.1f}")

print("Arriba del promedio:")
for banda, cantidad in votos:
    if cantidad > promedio:
      print(f"{banda} ({cantidad})")
```

```python
Ejercicio #3
paradas = (("Querétaro", 0), ("San Juan del Río", 52), ("Tequisquiapan", 74),
("Bernal", 125), ("Jalpan", 260))

tramo_mas_largo = ""
max_distancia = -1

for i in range(len(paradas) - 1):
    origen, km_origen = paradas[i]
    destino, km_destino = paradas[i+1]

    distancia = km_destino - km_origen
    print(f"{origen} : {destino}: {distancia} km")

    if distancia > max_distancia:
        max_distancia = distancia
        tramo_mas_largo = f"{origen}  {destino}"

print(f"Tramo más largo: {tramo_mas_largo} ({max_distancia} km)")
```

```python
#Ejercicio 4
temps = (8, 14, 22, 27, 31, 19, 11)

def clasificar(t):
    if t < 12:
        return "frio"
    elif 12 <= t <= 24:
        return "templado"
    else:
        return "calor"

climas = []
for t in temps:
    climas.append(clasificar(t))

print(climas)


conteo = {}
for climas in climas:
    if climas in conteo:
        conteo[climas] += 1
    else:
        conteo[climas] = 1

print(conteo)


predominante = ""
max_conteo = -1
for clima, cantidad in conteo.items():
    if cantidad > max_conteo:
        max_conteo = cantidad
        predominante = clima

print(f"Predomina: templado")
```

```python
#Ejercicio 5
setlist = (("Intro", 95), ("Riff asesino", 245), ("Balada", 302), ("Solo", 128), ("Encore", 409))

def formato(seg):
    min = seg // 60
    seg = seg % 60

    if seg < 10:
        seg = f"0{seg}"
    else:
        seg = str(seg)

    return f"{min}:{seg}"

total_segundos = 0
for cancion, duracion in setlist:
    total_segundos += duracion
    print(f"{cancion}: {formato(duracion)}")

print(f"Duración total: {formato(total_segundos)}")
```
