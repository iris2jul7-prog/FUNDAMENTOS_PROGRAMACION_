```python
#FUNCIONES DE TUPLAS: PARTE A
def sumar_tupla(tupla):
    total = 0
    for n in tupla:
        total = total + n
    return total

def seccion_tuplas():
    print(" TUPLAS ")
    numeros = (8, 3, 15, 1, 6)
    print("Tupla original:", numeros)
    print("Tercer elemento:", numeros[2])
    try:
        a = int(input("Escribe un número: "))
        b = int(input("Escribe otro número: "))
        nueva_tupla = numeros + (a, b)
        print("Nueva tupla:", nueva_tupla)
        print("La tupla original no cambia:", numeros)
        lista = list(nueva_tupla)
        lista.sort()
        print("Lista ordenada:", lista)
        print("Suma de la nueva tupla:", sumar_tupla(nueva_tupla))
    except ValueError:
        print("Error: Debes ingresar números enteros válidos.")
```

```python
#DICCIONARIOS PARTE B
def buscar_telefono(diccionario, nombre):
    if nombre in diccionario:
        return diccionario[nombre]
    else:
        return None

def seccion_diccionarios():
    print(" DICCIONARIOS ")
    contactos = {"Isa": "4434925449", "Ale": "4461319445", "Iris": "4428100553"}
    nombre = input("Nombre del nuevo contacto: ")
    telefono = input("Teléfono del nuevo contacto: ")
    contactos[nombre] = telefono

    print("Nombres en tus contactos:")
    for clave in contactos:
        print("-", clave)

    buscar = input("¿A quién quieres buscar? ")
    resultado = buscar_telefono(contactos, buscar)
    if resultado is None:
        print(buscar, "no está en tus contactos")
    else:
        print("El teléfono de", buscar, "es", resultado)
```

```python
#EXCEPCIONES PARTE C
def seccion_excepciones():
    print(" EXCEPCIONES ")
    try:
        num1 = int(input("Primer número entero: "))
        num2 = int(input("Segundo número entero: "))
        print("Suma:", num1 + num2)
        print("División:", num1 / num2)
    except ValueError:
        print("No valido: debes escribir un número entero, no letras ni dejarlo vacío.")
    except ZeroDivisionError:
        print("No valido: no se puede dividir entre cero, Intenta con otro número.")

```

```python
#STRINGS PARTE D
def contar_palabras(texto):
    palabras = texto.split()
    return len(palabras)

def seccion_strings():
    print(" STRINGS ")
    mensaje = "me gusta escuchar Taylor Swift "
    print("Mensaje:", mensaje)
    print("Longitud:", len(mensaje))
    print("Mayúsculas:", mensaje.upper())
    print("Reemplazo:", mensaje.replace("Taylor Swift", "Olivia Rodrigo"))
    print("Número de palabras:", contar_palabras(mensaje))
```

```python
#MENU PRINCIPAL PARTE E
def menu_principal():
    opcion = ""
    while opcion != "5":
        print()
        print("1) Tuplas 2) Diccionarios 3) Excepciones 4) Strings 5) Salir")
        opcion = input("Elige una opción: ")
        if opcion == "1":
            seccion_tuplas()
        elif opcion == "2":
            seccion_diccionarios()
        elif opcion == "3":
            seccion_excepciones()
        elif opcion == "4":
            seccion_strings()
        elif opcion == "5":
            print("Saliste del programa.")
        else:
            print("Opción inválida")
menu_principal()
```
